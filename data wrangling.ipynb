Load Space X dataset, from last section.


```python
df=pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
df.head(10)
```

```python
df.dtypes
```

```python
# Apply value_counts() on column LaunchSite
launch_site_counts = df['LaunchSite'].value_counts()
print(launch_site_counts)
```

```python
# Apply value_counts on Orbit column
orbit_counts = df['Orbit'].value_counts()
print(orbit_counts)

# Exclude GTO since it's a transfer orbit, not a destination orbit itself
orbit_counts_no_gto = df[df['Orbit'] != 'GTO']['Orbit'].value_counts()
print(orbit_counts_no_gto)
```

```python
# landing_outcomes = values on Outcome 
landing_outcomes = df['Outcome'].value_counts()
print(landing_outcomes)
```

```python
for i,outcome in enumerate(landing_outcomes.keys()):
    print(i,outcome)
```

```python
bad_outcomes=set(landing_outcomes.keys()[[1,3,5,6,7]])
bad_outcomes
```

```python
# landing_class = 0 if bad_outcome
# landing_class = 1 otherwise
bad_outcome = {'False ASDS', 'False Ocean', 'False RTLS', 'None ASDS', 'None None'}

# Create landing_class: 0 if outcome is in bad_outcome, else 1
landing_class = [0 if outcome in bad_outcome else 1 for outcome in df['Outcome']]

# Assign it back to the dataframe (optional but commonly done next)
df['Class'] = landing_class

print(landing_class[:10])  # preview first 10 values
```

This variable will represent the classification variable that represents the outcome of each launch. If the value is zero, the  first stage did not land successfully; one means  the first stage landed Successfully 


```python
df['Class']=landing_class
df[['Class']].head(8)
```
