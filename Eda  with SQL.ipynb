```python
import pandas as pd
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False,method="multi")
```

**Note:This below code is added to remove blank rows from table**


```python
#DROP THE TABLE IF EXISTS

%sql DROP TABLE IF EXISTS SPACEXTABLE;
```

```python
%sql create table SPACEXTABLE as select * from SPACEXTBL where Date is not null
```

```python
%%sql
SELECT DISTINCT "Launch_Site" FROM SPACEXTBL;
```

```python
%%sql
SELECT * FROM SPACEXTBL WHERE "Launch_Site" LIKE 'CCA%' LIMIT 5;
```

```python
%%sql
SELECT SUM("PAYLOAD_MASS__KG_") AS "Total_Payload_Mass_KG"
FROM SPACEXTBL
WHERE "Customer" = 'NASA (CRS)';
```

```python
%%sql
SELECT AVG("PAYLOAD_MASS__KG_") AS "Average_Payload_Mass_KG"
FROM SPACEXTBL
WHERE "Booster_Version" = 'F9 v1.1';
```

```python
%%sql
SELECT MIN("Date") AS "First_Successful_Ground_Pad_Landing"
FROM SPACEXTBL
WHERE "Landing_Outcome" = 'Success (ground pad)';
```

```python
%%sql
SELECT
    (SELECT COUNT(Mission_Outcome) FROM SPACEXTBL WHERE Mission_Outcome LIKE '%Success%') AS Successful_Mission_Outcomes,
    (SELECT COUNT(Mission_Outcome) FROM SPACEXTBL WHERE Mission_Outcome LIKE '%Failure%') AS Failed_Mission_Outcomes
```

```python
%%sql
SELECT DISTINCT Booster_Version, MAX(PAYLOAD_MASS__KG_) AS "Maximum Payload Mass"
FROM SPACEXTBL
GROUP BY Booster_Version
ORDER BY "Maximum Payload Mass" DESC
```

```python
%%sql
SELECT "Mission_Outcome", COUNT(*) AS "Total_Count"
FROM SPACEXTBL
GROUP BY "Mission_Outcome";
```

```python
%%sql
SELECT "Booster_Version"
FROM SPACEXTBL
WHERE "PAYLOAD_MASS__KG_" = (SELECT MAX("PAYLOAD_MASS__KG_") FROM SPACEXTBL);
```

```python
%%sql
SELECT
    CASE CAST(strftime('%m', Date) AS INTEGER)
        WHEN 1 THEN 'January' WHEN 2 THEN 'February' WHEN 3 THEN 'March'
        WHEN 4 THEN 'April' WHEN 5 THEN 'May' WHEN 6 THEN 'June'
        WHEN 7 THEN 'July' WHEN 8 THEN 'August' WHEN 9 THEN 'September'
        WHEN 10 THEN 'October' WHEN 11 THEN 'November' WHEN 12 THEN 'December'
    END AS Month,
    Booster_Version,
    Launch_Site,
    Landing_Outcome
FROM SPACEXTBL
WHERE Landing_Outcome LIKE '%Success%'
AND strftime('%Y', Date) = '2017'
```

```python
%%sql
SELECT DISTINCT Booster_Version, MAX(PAYLOAD_MASS__KG_) AS "Maximum Payload Mass"
FROM SPACEXTBL
GROUP BY Booster_Version
ORDER BY "Maximum Payload Mass" DESC
```

```python
%%sql
SELECT "Landing_Outcome", COUNT(*) AS "Outcome_Count"
FROM SPACEXTBL
WHERE "Date" BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY "Landing_Outcome"
ORDER BY "Outcome_Count" DESC;
```
